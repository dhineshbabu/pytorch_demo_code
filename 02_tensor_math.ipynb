```python
import torch
import numpy as np
```

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])

# tensor addition
tensor_c = tensor_a + tensor_b
tensor_c
```

```python
# we can also use torch.add method to add 2 tensors
torch.add(tensor_a, tensor_b)
```

```python
# subtraction
tensor_b - tensor_a
```

```python
# another way
torch.sub(tensor_a, tensor_b)
```

```python
# tensor multiplication
torch.mul(tensor_b, tensor_a)
```

```python
# another way for multiplication
torch.mul(tensor_a, tensor_b)
```

```python
# tensor division
tensor_b / tensor_a
```

```python
# another way
torch.div(tensor_b, tensor_a)
```

```python
# remainder / modulus
tensor_b % tensor_a
```

```python
# another way
torch.remainder(tensor_b, tensor_a)
```

```python
torch.remainder(tensor_a, tensor_b)
```

```python
# Exponential / power
torch.pow(tensor_a, tensor_b)
```

```python
# another shortway to achieve the operations
tensor_a.add(tensor_b)
```

```python
# short form of += is by adding _ Underscore to the method name like add_
tensor_a.add_(tensor_b) # now this is reflected in tensor_a
print(tensor_a)
```
